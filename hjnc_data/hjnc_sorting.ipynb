# HJNC 데이터 분류

**Github에 있는 'HJNC_2025_annual.csv'를 사용하였습니다.**

```python
csv_path = 'C:/Users/user/Desktop/data_practice/busan_project/hjnc_data/HJNC_2025_annual.csv'
df = pd.read_csv(csv_path)

df.head()
```

```python
df.info()
```

```python
df_2 = df.copy()

df_sorted = df_2[['컨테이너번호','T/S 구분', '반입', '반출', '반입시간', '반출시간', '모선항차', '차량번호', 'F/M', 'Class', 'Sz/Tp', '적하항', 'Booking번호', '봉인번호', ]]

df_sorted.head(10)
```

```python
print(df_sorted["T/S 구분"].value_counts())
```

```python
'''
타 터미널 웹사이트에서 조회할 컨테이너 정보들만 추렸습니다.
찾아보니 T/S 구분이 P라는 것은 'HJNC에서 들어와 타부두 환적을 한 경우'라고 합니다.
HJNC에서 타 터미널로 넘어간 컨테이너만 보고 싶기 때문에,
Vessel로 반입, Truck으로 반출되고
Class가 Import(수입)인 컨테이너로 범위를 좁혔습니다.
'''

df_v_to_t = df_sorted[
    (df_2["T/S 구분"] == "P") &
    (df_2['반입'] == 'Vessel')
]

df_v_to_t
```

```python
print(len(df_v_to_t))

# 컨테이너가 255968개로 좁혀졌습니다.
```
